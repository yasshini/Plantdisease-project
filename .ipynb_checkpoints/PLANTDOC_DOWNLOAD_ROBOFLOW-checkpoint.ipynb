# PLANT DISEASE DETECTION USING YOLOV8- Part I

Detection of Plant Diseases is a vital stage in the agricultural sector because it gives farmers to take timely diagnosis and maintain good yield of crops, which in turn affect agricultural productivity on which economy higly depends on. Manual detection of plant diseases is not a viable option and since many deep learning techniques have emerged over the years, these automated systems have made the detection task much faster and easy to use.

The YOLO deep learning model which is a Single Shot Detector (SSD) is gaining prominence over recent years in various object detection, classification and segmentation tasks giving accurate results quickly. YOLO models work by dividing the image into a grid of S x S dimensions and then assigning probabilities for a bounding box to occur in a cell in the grid followed by the Non Maximal Suppression technique which selects the best bounding box for the detected Region of Interest (ROI). YOLOv8 is the latest version of YOLO by Ultralytics which builds on the success of previous versions, introducing new features and improvements for enhanced performance, flexibility, and efficiency.

Previous work done in the same domain by using older YOLO version gave promising results with a lot of room for improvement in the model performance and since no relevant work has been published so far for plant disease detection using YOLOv8, hence this project aims to use the latest state of the art model to detect and classify Plant Diseases quickly and accurately over various classes. For training the model the PlantDoc dataset is used, whose information is given in sections below.

There are two notebooks for this task, the first is downloading the dataset  and the second is training the YOLOv8 model using pre-trained weights whose details are explained in the respective notebook.



# DOWNLOAD DATASET

This code installs the pre-processed PlantDoc Dataset from the Roboflow platform.

The PlantDoc dataset was originally published by researchers at the Indian Institute of Technology, and described in depth in their [paper](https://arxiv.org/pdf/1911.10317.pdf). One of the paper’s authors, Pratik Kayal, shared the object detection dataset available on [GitHub](https://github.com/pratikkayal/PlantDoc-Dataset)

PlantDoc is a dataset of 2,569 images across 13 plant species and 30 classes (diseased and healthy) for image classification and object detection. There are 8,851 labels. Read more about how the version available on Roboflow improves on the original version [here](https://blog.roboflow.ai/introducing-an-improved-plantdoc-dataset-for-plant-disease-object-detection/).
